# 00: Look at the data

Before any model: what did the class record, and are your own takes good?
Run all cells, then look at the plots.

```python
import numpy as np
import matplotlib.pyplot as plt
from collections import Counter
from load import load, features, GESTURES, FEATURE_NAMES   # load.py next to this notebook

# Download the class data once, so your numbers stay comparable while others keep recording:
#   curl -o gestures.zip "<export link from the projector>"
SRC = "gestures.zip"
X, y, person = load(SRC)
people = np.unique(person)
print(X.shape, "= takes x time steps (50 Hz) x [acc_x acc_y acc_z gyro_x gyro_y gyro_z]")
```

## Who recorded what

```python
c = Counter(zip(person, y))
w = max(len(p) for p in people)
print("person".ljust(w), *[g.rjust(7) for g in GESTURES])
for p in sorted(people, key=str.lower):
    print(p.ljust(w), *[str(c[(p, g)]).rjust(7) for g in GESTURES])
```

## Your own takes
Flat lines, clipped values, or plainly the wrong gesture? Note them down; your instructor can remove them.

```python
ME = "your name as typed on the phone"

t = np.arange(X.shape[1]) / 50
mine = person == ME
fig, axs = plt.subplots(2, 6, figsize=(20, 6), sharex=True)
for j, g in enumerate(GESTURES):
    for i in np.where(mine & (y == g))[0]:
        axs[0, j].plot(t, X[i, :, :3], lw=0.7)
        axs[1, j].plot(t, X[i, :, 3:], lw=0.7)
    axs[0, j].set_title(f"{g} ({(mine & (y == g)).sum()} takes)")
axs[0, 0].set_ylabel("acc x,y,z  m/s²"); axs[1, 0].set_ylabel("gyro x,y,z  rad/s"); axs[1, 0].set_xlabel("s")
plt.tight_layout()
```

## One gesture, different people
Same label, different styles: speed, size, grip.

```python
g = "left"
ps = people[:6]
fig, axs = plt.subplots(2, len(ps), figsize=(4 * len(ps), 6), sharex=True, squeeze=False)
for k, p in enumerate(ps):
    for i in np.where((person == p) & (y == g))[0][:5]:
        axs[0, k].plot(t, X[i, :, :3], lw=0.7)
        axs[1, k].plot(t, X[i, :, 3:], lw=0.7)
    axs[0, k].set_title(f"{p}: {g}")
axs[0, 0].set_ylabel("acc"); axs[1, 0].set_ylabel("gyro")
plt.tight_layout()
```

## Where is gravity?
The mean acceleration of a take is mostly gravity: it tells you how the phone was held. Upright, screen facing you: y ≈ +9.8.

```python
g_mean = X[:, :, :3].mean(1)
for p in sorted(people, key=str.lower):
    print(f"{p:>{w}}  mean acc x,y,z = {np.round(g_mean[person == p].mean(0), 1)}")
```
